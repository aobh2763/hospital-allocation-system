```python
import numpy as np
import pandas as pd

from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent.parent
DATA_DIR = PROJECT_ROOT / "data"

PATIENT_DATA_PATH = DATA_DIR / "patients"
HOSPITAL_DATA_PATH = DATA_DIR / "hospital"

print(DATA_DIR)
```

```python
df = pd.read_csv(DATA_DIR / "patients.csv")

df.head()
```

```python
pd.Series(df.columns, name="column").to_csv(DATA_DIR / "output" / "patient_columns.csv", index=False)
```

```python
df["age"].describe()
```

```python
step = 1/12

df["age_years"] = np.floor(df["age"]).astype(int)
df["age_rest"] = df["age"] - df["age_years"]
df["age_months"] = np.floor(df["age_rest"] / step).astype(int)

df.drop(columns=["age", "age_rest"], inplace=True)
```

```python
df[["age_years", "age_months"]].describe()
```

```python
df.head(5)
```

```python
df["GENDER"].value_counts()
```

```python
df["gender_m"] = (df["GENDER"] == "M").astype(int)
df["gender_f"] = (df["GENDER"] == "F").astype(int)

df.drop(columns=["GENDER"], inplace=True)

df.head(5)
```

```python
df["RACE"].value_counts()
```

```python
race_dict = {
    "white": 0,
    "black": 1,
    "asian": 2,
    "other": 3,
    "hawaiian": 4,
    "native": 5
}

df["race"] = df["RACE"].map(race_dict)
df.drop(columns=["RACE"], inplace=True)

df.head(5)
```

```python
df["ETHNICITY"].value_counts()
```

```python
df["is_hispanic"] = (df["ETHNICITY"] == "hispanic").astype(int)
df.drop(columns=["ETHNICITY"], inplace=True)

df.head(5)
```

```python
df_missing_columns = df.isnull().sum().to_frame(name="missing_count")
df_missing_columns["missing_percentage"] = df_missing_columns["missing_count"] / len(df) * 100
df_missing_columns.to_csv(DATA_DIR / "output" / "patient_missing_columns.csv")

missing_threshold = 50

dropped_columns = df_missing_columns[df_missing_columns["missing_percentage"] > missing_threshold].index
df.drop(columns=dropped_columns, inplace=True)
df_missing_columns.loc[dropped_columns].to_csv(DATA_DIR / "output" / "patient_dropped_columns.csv")
```

```python
obs_cols = [c for c in df.columns if c.startswith("obs_")]

bins = [-1, 1, 5, 12, 18, 30, 45, 65, 80, 200]
df["age_grp"] = pd.cut(df["age_years"], bins)

grp_med = df.groupby("age_grp", observed=True)[obs_cols].transform("median")
df[obs_cols] = df[obs_cols].fillna(grp_med).fillna(df[obs_cols].median())

df = df.drop(columns="age_grp")
print(df[obs_cols].isna().sum().sum())
```

```python
df = df[[c for c in df.columns if c not in ["risk_score", "risk_class"]] + ["risk_score", "risk_class"]]
```

```python
risk_to_val = {
    "Low": 0,
    "Medium": 1,
    "High": 2,
    "Critical": 3
}

df["risk_class"] = df["risk_class"].map(risk_to_val)
```

```python
df.to_csv(DATA_DIR / "patient_preprocessed.csv", index=False)
```
